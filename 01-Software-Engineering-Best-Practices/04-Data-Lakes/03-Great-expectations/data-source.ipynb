### Example validate column names

Below is example of how to setup a validation by running `validator.method` it sets up the validator to expect that thing run that cell and the move on!

````python
columns = ["rank", "title", "site", "link", "score", "author", "comments_number"]
for col in columns:
    validator.expect_column_to_exist(col)
````

# Setup validations for each column

Now you need to setup the validations for the all of the columns using the syntax 


```python
validator.method
```

and finding the correct method in order to meet the requirements of the list!

### Validate rank

- Type of rank in int64
- Rank is unique
- Rank is not less than 1

````python
# Validate rank

# Tipo int64
validator.expect_column_values_to_be_of_type("rank", "int64")

# Rank es único
validator.expect_column_values_to_be_unique("rank")

# Rank no es menor que 1
validator.expect_column_min_to_be_between("rank", min_value=1)
````

### Validate Title 

- Type of string
- Can't be null
- Lengths should be between 1 and 255

````python
# Validate Title

# Tipo string (en pandas suele ser 'object')
validator.expect_column_values_to_be_of_type("title", "object")

# No puede ser nulo
validator.expect_column_values_to_not_be_null("title")

# Longitud entre 1 y 255 caracteres
validator.expect_column_value_lengths_to_be_between(
    "title", min_value=1, max_value=255
)
````

### Validate Site 

- Type of string
- Can't be null
- Should match the regex of a valid domain

````python
# Validate Site

# Tipo string
validator.expect_column_values_to_be_of_type("site", "object")

# No puede ser nulo
validator.expect_column_values_to_not_be_null("site")

# Debe seguir un patrón de dominio (ej. example.com, news.ycombinator.com, etc.)
validator.expect_column_values_to_match_regex(
    "site",
    r"^[a-zA-Z0-9.-]+\.[a-zA-Z]{2,}$"
)
````

### Validate Link

- Type of string
- Can't be null
- Should match the regex of the a valid link

````python
# Validate Link

# Tipo string
validator.expect_column_values_to_be_of_type("link", "object")

# No puede ser nulo
validator.expect_column_values_to_not_be_null("link")

# Debe parecer un link http/https
validator.expect_column_values_to_match_regex(
    "link",
    r"^https?://.+$"
)
````

### Validate Score

- Type of int64
- Must be atleast 0

````python
# Validate Score

# Tipo int64
validator.expect_column_values_to_be_of_type("score", "int64")

# Mínimo 0
validator.expect_column_min_to_be_between("score", min_value=0)
````

### Validate Author

- Type of string
- Can't be null

````python
# Validate Author

# Tipo string
validator.expect_column_values_to_be_of_type("author", "object")

# No puede ser nulo
validator.expect_column_values_to_not_be_null("author")
````

### Validate Comments Number

- Type of int64
- Must be zero or greater

````python
# Validate Comments Number

# Tipo int64
validator.expect_column_values_to_be_of_type("comments_number", "int64")

# Debe ser 0 o mayor
validator.expect_column_min_to_be_between("comments_number", min_value=0)

````
