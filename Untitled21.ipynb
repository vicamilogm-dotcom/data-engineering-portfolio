### 1. Fase de Ingesta y Limpieza (Data Wrangling)
En este paso preparamos los datos para que la red neuronal pueda procesarlos correctamente.

```python
import pandas as pd

# 1. Asegurar formato de fecha
# Usamos iloc[:, 3] según la estructura previa, o el nombre de columna si lo prefieres
df['f120_fecha'] = pd.to_datetime(df.iloc[:, 3])

# 2. Tratamiento de Devoluciones / Consumo Neto
# Agrupamos por referencia y fecha para consolidar el movimiento diario neto
# Esto es importante para que las devoluciones (negativos) se resten de las salidas
df_diario = df.groupby(['f120_referencia', 'f120_fecha']).sum(numeric_only=True).reset_index()

# 3. Ordenar cronológicamente para series de tiempo
df_diario = df_diario.sort_values(by=['f120_referencia', 'f120_fecha'])

print("Datos procesados (Nivel Diario):")
display(df_diario.head())
print(f"Total de registros: {len(df_diario)}")
```

### 2. Ingeniería de Características (Feature Engineering)
Creamos variables temporales, retardos (lags) y promedios móviles para ayudar al modelo a predecir mejor.

```python
# 1. Características de Tiempo
df_diario['mes'] = df_diario['f120_fecha'].dt.month
df_diario['dia_semana'] = df_diario['f120_fecha'].dt.dayofweek

# 2. Creación de Lags (Valores pasados)
# Agrupamos por referencia para que el lag no mezcle productos distintos
df_diario['lag_1'] = df_diario.groupby('f120_referencia')['salidas'].shift(1)
df_diario['lag_7'] = df_diario.groupby('f120_referencia')['salidas'].shift(7)

# 3. Medias Móviles (Ventana de 7 días)
df_diario['media_movil_7'] = df_diario.groupby('f120_referencia')['salidas'].transform(lambda x: x.rolling(window=7).mean())

# Eliminar filas con NaN resultantes de los lags/medias móviles
df_diario = df_diario.dropna()

print("Características generadas:")
display(df_diario.head())
print(f"Registros después de limpiar NaNs: {len(df_diario)}")
```

### 3. Preparación para LSTM (Escalado y Ventanas)
Para que la LSTM aprenda, necesitamos escalar los valores y transformar la tabla en secuencias temporales.

```python
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
import numpy as np

# 1. Asegurar que las referencias sean todas strings para evitar errores de tipo mixto
df_diario['f120_referencia'] = df_diario['f120_referencia'].astype(str)

# 2. Codificación de Referencias
le = LabelEncoder()
df_diario['ref_encoded'] = le.fit_transform(df_diario['f120_referencia'])

# 3. Normalización de variables numéricas
features_to_scale = ['salidas', 'lag_1', 'lag_7', 'media_movil_7']
scaler = MinMaxScaler(feature_range=(0, 1))
df_diario[features_to_scale] = scaler.fit_transform(df_diario[features_to_scale])

def create_sequences(data, seq_length):
    """Convierte el DataFrame en secuencias para LSTM"""
    X, y = [], []
    # Nota: Por el volumen de datos (1.8M), esto consume mucha RAM.
    # Procesaremos por grupos o tomaremos una muestra si es necesario más adelante.
    for i in range(len(data) - seq_length):
        X.append(data[i:(i + seq_length)])
        y.append(data[i + seq_length, 0]) # Predice la columna 'salidas'
    return np.array(X), np.array(y)

print("Datos normalizados y referencias codificadas (corregido):")
display(df_diario.head())
```

### 4. Estrategia de Entrenamiento Cronológico
Dividimos la historia en:
*   **Base (2019-2023):** Para que el modelo aprenda el comportamiento general.
*   **Fine-Tuning (2024):** Para ajustar el modelo a la tendencia más reciente del histórico.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# Definir fecha de corte para validación (ejemplo: últimos 30 días)
fecha_corte = df_diario['f120_fecha'].max() - pd.Timedelta(days=30)

train_df = df_diario[df_diario['f120_fecha'] <= fecha_corte]
test_df = df_diario[df_diario['f120_fecha'] > fecha_corte]

# Seleccionar columnas para el modelo (X)
columnas_feat = ['ref_encoded', 'mes', 'dia_semana', 'lag_1', 'lag_7', 'media_movil_7']
target = 'salidas'

print(f"Registros de entrenamiento: {len(train_df)}")
print(f"Registros de prueba: {len(test_df)}")
```

### 5. Definición del Modelo LSTM
Configuramos una arquitectura base para procesar las secuencias temporales.

```python
def build_lstm_model(input_shape):
    model = Sequential([
        LSTM(64, return_sequences=True, input_shape=input_shape),
        Dropout(0.2),
        LSTM(32, return_sequences=False),
        Dropout(0.2),
        Dense(16, activation='relu'),
        Dense(1) # Predicción de 'salidas'
    ])

    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model

# Definimos longitud de la secuencia (ventana de tiempo, ej: 7 días)
n_steps = 7
n_features = len(columnas_feat)

model = build_lstm_model((n_steps, n_features))
model.summary()
```

### 6. Generador de Datos y Entrenamiento
Usaremos `TimeseriesGenerator` de Keras para manejar el volumen de datos de forma eficiente sin agotar la memoria RAM.

```python
from tensorflow.keras.preprocessing.sequence import TimeseriesGenerator

# Parámetros del generador
batch_size = 256

# Preparar matrices para el generador
X_train = train_df[columnas_feat].values
y_train = train_df[target].values

X_test = test_df[columnas_feat].values
y_test = test_df[target].values

# Crear generadores para entrenamiento y validación
train_generator = TimeseriesGenerator(X_train, y_train, length=n_steps, batch_size=batch_size)
test_generator = TimeseriesGenerator(X_test, y_test, length=n_steps, batch_size=batch_size)

# Configurar Early Stopping para evitar overfitting
callback = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

print("Generadores configurados. Iniciando entrenamiento...")

history = model.fit(
    train_generator,
    epochs=10,
    validation_data=test_generator,
    callbacks=[callback]
)
```

### 9. Ejecución Integral: Definición, Entrenamiento y Guardado
Esta celda asegura que el modelo se cree, se entrene con los datos históricos y se guarde en Drive en un solo paso.

```python
import pandas as pd
import numpy as np
import joblib
import os
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
from tensorflow.keras.preprocessing.sequence import TimeseriesGenerator

# 1. CARGA Y WRANGLING (Aseguramos que df_diario exista)
print("1/4 Cargando y procesando datos originales...")
file_path = '/content/drive/MyDrive/data/mp_historico_consolidado.csv'
df = pd.read_csv(file_path, low_memory=False)
df['f120_fecha'] = pd.to_datetime(df.iloc[:, 3])
df_diario = df.groupby(['f120_referencia', 'f120_fecha']).sum(numeric_only=True).reset_index()
df_diario = df_diario.sort_values(by=['f120_referencia', 'f120_fecha'])

# 2. INGENIERÍA DE CARACTERÍSTICAS
print("2/4 Generando características y escalando...")
df_diario['mes'] = df_diario['f120_fecha'].dt.month
df_diario['dia_semana'] = df_diario['f120_fecha'].dt.dayofweek
df_diario['lag_1'] = df_diario.groupby('f120_referencia')['salidas'].shift(1)
df_diario['lag_7'] = df_diario.groupby('f120_referencia')['salidas'].shift(7)
df_diario['media_movil_7'] = df_diario.groupby('f120_referencia')['salidas'].transform(lambda x: x.rolling(window=7).mean())
df_diario = df_diario.dropna()

# Encoder y Scaler
le = LabelEncoder()
df_diario['f120_referencia'] = df_diario['f120_referencia'].astype(str)
df_diario['ref_encoded'] = le.fit_transform(df_diario['f120_referencia'])

features_to_scale = ['salidas', 'lag_1', 'lag_7', 'media_movil_7']
scaler = MinMaxScaler(feature_range=(0, 1))
df_diario[features_to_scale] = scaler.fit_transform(df_diario[features_to_scale])

# 3. SEGMENTACIÓN Y GENERADORES
columnas_feat = ['ref_encoded', 'mes', 'dia_semana', 'lag_1', 'lag_7', 'media_movil_7']
target = 'salidas'
n_steps, batch_size = 7, 256

train_base_df = df_diario[df_diario['f120_fecha'].dt.year <= 2023]
ft_2024_df = df_diario[df_diario['f120_fecha'].dt.year == 2024]

def get_gen(data):
    X, y = data[columnas_feat].values, data[target].values
    return TimeseriesGenerator(X, y, length=n_steps, batch_size=batch_size)

base_gen = get_gen(train_base_df)
ft_2024_gen = get_gen(ft_2024_df)

# 4. MODELO, ENTRENAMIENTO Y GUARDADO
print("3/4 Entrenando modelo (Base + Fine-Tuning)...")
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(n_steps, len(columnas_feat))),
    tf.keras.layers.LSTM(64, return_sequences=True),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.LSTM(32, return_sequences=False),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1)
])
model.compile(optimizer='adam', loss='mse', metrics=['mae'])

model.fit(base_gen, epochs=5, verbose=1) # Reducido a 5 epochs para rapidez
model.optimizer.learning_rate = 1e-4
model.fit(ft_2024_gen, epochs=3, verbose=1)

print("4/4 Guardando resultados en Drive...")
model_path = '/content/drive/MyDrive/data/models/'
os.makedirs(model_path, exist_ok=True)
model.save(f'{model_path}demand_lstm_v1.keras')
joblib.dump(scaler, f'{model_path}scaler_v1.pkl')
joblib.dump(le, f'{model_path}label_encoder_v1.pkl')

print(f"\n✅ ¡TODO LISTO! Modelo y artefactos guardados en: {model_path}")
```

Graficamos la proyección de 90 días para detectar anomalías o tendencias.

```python
def predict_horizon_advanced(model, start_sequence, horizon=90):
    current_seq = start_sequence.copy()
    predictions = []

    for _ in range(horizon):
        # Predicción
        pred = model.predict(current_seq.reshape(1, n_steps, len(columnas_feat)), verbose=0)[0, 0]
        predictions.append(pred)

        # Crear nueva fila basada en la última
        last_row = current_seq[-1].copy()

        # Actualizar Lags recursivamente
        # lag_1 es la predicción anterior
        # lag_7 es lo que estaba en la posición 7 hace un paso
        new_lag_1 = pred
        new_lag_7 = current_seq[-7, 3] if len(current_seq) >= 7 else pred

        # Actualizar Media Móvil simple (promedio de los últimos 7 valores predichos/reales)
        recent_values = np.append(current_seq[:, 0], pred)
        new_rolling_7 = np.mean(recent_values[-7:])

        # Construir nueva fila: ['ref_encoded', 'mes', 'dia_semana', 'lag_1', 'lag_7', 'media_movil_7']
        # Nota: 'mes' y 'dia_semana' deberían avanzar en el tiempo idealmente,
        # aquí mantenemos los del último paso para simplificar o podrías incrementarlos.
        new_row = np.array([last_row[0], last_row[1], (last_row[2]+1)%7, new_lag_1, new_lag_7, new_rolling_7])

        # Desplazar ventana
        current_seq = np.append(current_seq[1:], [new_row], axis=0)

    return np.array(predictions)

# Ejecutar nueva proyección
ultima_ventana_data = ft_2024_df[columnas_feat].values[-n_steps:]
proyeccion_scaled = predict_horizon_advanced(model, ultima_ventana_data)

# Des-escalar
dummy = np.zeros((len(proyeccion_scaled), len(features_to_scale)))
dummy[:, 0] = proyeccion_scaled
proyeccion_real = scaler.inverse_transform(dummy)[:, 0]

# Graficar con más detalle
plt.figure(figsize=(14, 6))
plt.plot(proyeccion_real, color='red', linewidth=2, label='Demanda Proyectada 2025')
plt.axhline(y=np.mean(proyeccion_real), color='gray', linestyle='--', label='Media de Proyección')
plt.title('Pronóstico Dinámico de Demanda (Enero-Marzo 2025)')
plt.xlabel('Días desde el inicio de 2025')
plt.ylabel('Cantidad de Unidades')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
